<b>Real-World Challenge &ndash; The SCSU School Closing Calculator!

SCSU, like most universities, will close campus and cancel/go online with in-person classes if the weather is going to be bad.  SCSU was notoriously bad at this for a while; a couple years ago, the administration often waited until late evening to cancel in-person classes. In an endeavor to help the poor folks who couldn't seem to make this call sooner in the day, let's make a program to help them!

First, we need to establish if the temperature outside is too cold to open, if it isn't, then we'll check the wind chill and see if that's too cold.  

For temperature, we'll use a threshold of -25 &deg;F as some cars may struggle to start at this temperature and frostbite can begin to set in on exposed skin in less than 30 minutes.  For wind chill, we'll use a threshold of -35 &deg;F, since skin is prone to frostbite in 10 minutes under those conditions.

Hint: Think about this logically &ndash; if it's too cold to open, we don't even need to check the wind chill because campus already should close.  (This might help you with where to calculate WC in your function.)

We need to calculate wind chill because wind chill is derived from observed temperature (T) and wind speed (v, for velocity).  The formula to calculate wind chill (WC) using Fahrenheit for T and miles per hour for v is:

$$ \large WC = 35.74 + (0.6215 * T) - (35.75 * v^{0.16}) + (0.4275 * T * v^{0.16}) $$

<i>6) Write code to check if SCSU should open! Your program should:
- <i>read in two variables as inputs (T for temperature and v for wind speed).
- <i>use if statements to determine if St. Cloud State should close due to dangerously cold temperature or wind chill.  To close,
  - <i>temperature must be colder than or equal to -25 &deg;F.
  - <i>wind chill must be colder than or equal to -35 &deg;F.
- <i>print a response about whether SCSU should open or not. Your print statement should include WHY the decision should be to close (i.e., SCSU should close because the wind chill will be -40.).</i>

<i>Test this code with three cases that should cover every outcome:
- T = -26, v = 0 (close for cold)
- T = -23, v = 20 (close for wind chill)
- T = 10, v = 15 (should not close)
</i>

```python
#6 (T = -26, v = 0, WC = 19.581) (T = -23, v = 20, WC = -52.16892767) (T = 10, v = 15, WC = -6.589534421)
T = int(input())
v = int(input())

WC = 35.74 + (0.6215 * T) - (35.75 * v ** 0.16) + (0.4275 * T * v ** 0.16)

if T <= -25:
    print(f"SCSU should close because the temperature will be {T}")
elif WC <= -35:
    print(f"SCSU should close because the wind chill will be {WC}")
else:
    print("SCSU should not close because the temperature is not cold enough")
    
```
