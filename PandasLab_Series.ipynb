Pandas is the Python Data Analysis Library. The primary data structures in Pandas are the Series and the DataFrame. A Series is one-dimensional, while a DataFrame is two-dimensional. NumPy, Pandas, and Matplotlib all work very well together when working with real data.  
  
A few notes about Pandas (and why I use it frequently in my programs):
- It allows more than one data type in a Series or DataFrame (unlike NumPy arrays), making it ideal for numerical data as well as categorical/string data
- DataFrames are very similar in pandas to dataframes in R
- It allows the use of headers and column labels (unlike NumPy arrays)
- Simple handling of missing data in files
- Because you can have headers, you can slice the data based on the labels instead of numbers (but you can still use numbers if you want!)
- Allows extremely easy ways to create a time-series with data


<b>Part 1: Introducing the Series

Series are one-dimensional arrays of data and are always capitalized when referring to the Pandas data object.  I'll keep consistent there as much as possible; however, if I refer to a random series of data that's not specifically in Pandas, I'll use the lowercase version for clarity.  

Let's import the necessary libraries for today:

```python
import numpy as np
import pandas as pd
import random  # We'll use this later...
```

Pandas Series are one-dimensional, which means they're the equivalent of a vector in math/matrix world or a single column (or row) of data in Excel/spreadsheet world.  Series can be very helpful for individual runs of data.  Let's make a couple simple Series, then talk through what's going on in each:

```python
indices=['A','B','C','D','F']
amandavalues=[92,84,76,68,0]
alanvalues=[90,80,70,60,0]
amandagradebook=pd.Series(data=amandavalues, index=indices)
alangradebook=pd.Series(alanvalues, indices)
print(amandagradebook)
print(alangradebook)
```

There's a few things going on here.  First, let's talk about the indices (the column on the left).  Each index value can be specified as a list as long as it's the same length as the data that goes along with it.  In this case, we have five elements in both our data and our index list, so we're good there.  When printing a series, the indices always show up on the left.  

Now, let's also look at the list of amandavalues.  When we create the series based on this list, Pandas will take and pair each index to its corresponding data value.  Thus, A is associated with 92, B with 84, etc.  For this first Series (amandagradebook), we hard-coded the keywords with their corresponding input data (like specifying data=amandavalues).  

However, we don't always need to do this because Python can usually handle positional arguments (like assuming the data comes first in the variable list).  This was done in the creation of alangradebook.  However, even with this, I recommend hard-coding each keyword &ndash; it just makes your code easier to work with.  Let's see why this could be a problem (or at least confusing):

<i>1) If using positional keywords when defining a Series, does the order of the keywords matter if they're hard-coded (like explicitly stating data=alanvalues)?  Run the sample below to help:

```python
#1
q1test=pd.Series(index=indices, data=amandavalues)
print(q1test)
# Put your answer below here as a comment:
# No the order of hard-coded keywords does not matter. Python matches each value to its keyword regardless of order.
```

<i>2) Try the same, but without the keywords.  Does this work as expected/is it the same as the last question?  Why or why not?  What did Pandas do?

When creating Series (or any other Python data object), it's always helpful to check to make sure that Python is doing what you think it's doing.  My code is full of print statements to check things, especially in preliminary development.  

Now, let's look at one reason why Series are so useful:

```python
print(amandagradebook['B'])
```

We could always get to that the other way too:

```python
print(amandagradebook[1])
```

But then we'd have to know which index (by number position) for each call we'd make &ndash; and they're changing this type of pulling a value anyway.  I find the named indices much more useful.  Thus, we can easily use the named (or otherwise specified) indices to retrieve whatever value we want that corresponds with that index.  This can be very helpful when coding!  

<b>Part 2: Dog Heights?

In some ways, the process we just went through (with indices and values) should feel similar to the dictionary datatype we learned about earlier with keys and values.  Here, instead, we have indices and values.  Now, let's apply that to some data about dog sizes.

Making a Series out of a dictionary is super easy &ndash; you only need to put in the name of the dictionary in the Series function call and it takes care of the rest.  Let's try it using data from https://www.entirelypets.com/dog-breed-guide.html.  (Hint: pull this site up; we'll need it later.)  

We'll use the shortest average height of a few different dog breeds to create our dictionary.

```python
dogheightdict={'Chihuahua':6, 'Pomeranian':8,  'Bichon Frise':9, 'French Bulldog':12, 'Shetland Sheepdog':13, 'Collie':18}
print(type(dogheightdict))
dogheightdict
```

So, this output shows as a dictionary (notice the output has the curly brackets).  Now, let's make the Series:

```python
dogheights = pd.Series(dogheightdict)
dogheights
```

Modifying a Series is fairly easy as well; we just need to create a new series with our new data and indices and combine them.  To do that, use the pandas.concat() function (in our case, the pd.concat() function).  More information on that here:<br>
https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.concat.html

<i>4) Create a new Series called newdogs that contains two more dog breeds and their shortest average heights &ndash; Basset Hounds and Great Danes.   (You don't need to make the dictionary first if you don't want to; you can just add the info as lists of data and indices into the new Series.)  

```python
#4
newdogs = pd.Series(data=[15,28], index=['Basset Hound', 'Great Dane'])
```

<i>5) Now, combine the two Series together using the pd.concat() function; call your new output Series moredogheights.  Hint: make sure you input a list of your existing Series to the pd.concat() function...

```python
#5
moredogheights = pd.concat([dogheights, newdogs])
moredogheights
```

If you read further on the concat() help page linked above, there's a useful keyword called verify_integrity.  This will make sure that two indices don't have the same value.  It's usually worth including this flag if you don't know if indices might overlap; for this problem, we don't have to worry much about it.  If you're dealing with lots of data, it might be worth including. However, it also may slow down your code.

I usually like to work with sorted lists and Series when possible.  Python doesn't care, but it helps me out.  We can use the sort_values() to help us sort.  More on that here:<br>

Let's first sort based on the values themselves (axis=0 because there's only one axis in a Series).  [More on axes in the next lab.]

<i>10) Use index labels to retrieve a subset Series called oneworddogs that only contains Chihuahua, Pomeranian, and Collie dog breeds (and their associated minimum average heights.)  
    
Hints:  You can use a for loop to loop through series (either data or index). You will need to search the index name for spaces. I would also recommend creating lists of names and the associated heights which you can append. Then combine the lists into a series at the end. 

```python
#10
names = []
heights = []

for breed in moredogheights.index:
    if ' ' not in breed:
        names.append(breed)
        heights.append(moredogheights[breed])

oneworddogs = pd.Series(data=heights, index=names)
print(oneworddogs)
```
