For this first part, we're going to run through a scenario that you're familiar with &ndash; grading assignments in a class.  Because we haven't learned about advanced Python variable types yet, we're going to keep the class size very small (four students) so we don't have to type so much.

Let's say we have four students in our class: Alex, Bernie, Chris, and Dana.  We want to make a function that can read in each of their scores and will return the average.  Let's just go with a simple version that's already written out for you:

```python
def avgScore(alexscore,berniescore,chrisscore,danascore):
    avg=(alexscore+berniescore+chrisscore+danascore)/4
    return avg
```

The above function is pretty simple &ndash; it adds four numbers together and divides by four to get the average.  Note that this function isn't really expandable, though.  If a new student were to join the class, we'd have to rewrite the whole function &ndash; both the list of arguments and the calculation.  But, for now, this will be good enough.  

Let's try running it, just putting in scores for each into the function call:

```python
avgScore(80,75,83,91)
```

You should have gotten an average of 82.25 returned by Python.  That's fine on it's own, but we can do something better.  We can store this as its own variable for use later!

```python
assignment1=avgScore(80,75,83,91)
assignment1
```

Now, we have a variable called assignment1 that has 82.25 stored in it.  

What if there was a second assignment?  Let's run something like that:

```python
assignment2=avgScore(71,68,80,63)
```

Note that there's no visible output from this code cell.  Because we didn't tell Python to output the contents of assignment2, it didn't return anything to the screen.  However, the answer is still stored in the variable assignment2 for use later:

```python
print("The average for assignment 1 was "+str(assignment1)+"; the average for assigment 2 was "+str(assignment2)+".")
```

The key result here is that we can use a function multiple times and store the output from each run into a different variable.  

We can also use variables as input to a function, like:

```python
alex3=95
bernie3=92
chris3=96
dana3=95
assignment3=avgScore(alex3,bernie3,chris3,dana3)
print(assignment3)
```

Python will complain if you try to put a string into a calculation, so do be careful of that.  (It'll throw enough warning/error messages at you that you'll know something went wrong!)

<b>The you-just-bought-a-house-and-now-want-to-paint-a-room problem!

So I bought a house a couple years ago and ran into this issue; you might run into the same soon enough...

In your new house, you decide to paint a room &ndash; but you don't know how much paint to buy!  The handy function you're going to write will help you solve that problem.  You're going to calcuate the area of all four walls combined as well as the area of the ceiling.  For ease of estimation, don't worry about windows, doors, etc., for this problem.

<i>3) Write a function called paintEstimator that:
- <i> takes three inputs (length, width, and height of room &ndash; we'll assume a rectangular room and flat ceiling)
- <i> calculates the square footage of the ceiling and the four walls (not painting the floor)
- <i> calculates the minimum number of gallons of ceiling paint and wall paint to buy (assume single coat of paint)
- <i> uses that information to print two sentences with the square footage and number of gallons required
 - <i>one sentence for the ceiling
 - <i>one sentence for the walls (combined)

<i> For this problem, assume a gallon of paint covers 400 square feet.

```python
#3
import math

def paintEstimator(length, width, height):
    ceiling = length * width
    wall_1 = (length * height) * 2
    wall_2 = (width * height) * 2
    walls = wall_1 + wall_2
    gallons_w = math.ceil(walls / 400)
    gallons_c = math.ceil(ceiling / 400)
    
    print(f"For {ceiling} sq ft of ceiling, you need a minimum of {gallons_c} gallons of paint.")
    print(f"For {walls} sq ft of walls, you need a minimum of {gallons_w} gallons of paint.")
    print()
    
```

<i> 4) Finally, test your function on three rooms:
- <i> Room 1 is 15 ft long x 12 ft wide x 9 ft high
- <i> Room 2 is 10 ft long x 10 ft wide x 8 ft high
- <i> Room 3 is 25 ft long x 15 ft wide x 9 ft high

```python
#4
paintEstimator(15, 12, 9)
paintEstimator(10, 10, 8)
paintEstimator(25, 15, 9)
```
