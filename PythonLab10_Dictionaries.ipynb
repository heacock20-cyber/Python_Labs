```python
#1
heenen = {
    "Alarm clock": "A small device designed to interrupt the best dream you have had all week at the exact moment it was getting good.",
    "Meeting": "A social ritual in which several people sit together to discuss scheduling another meeting, typically resolving nothing.",
    "Diet": "A brief period of optimism that begins on a Monday and ends quietly by Wednesday when someone brings donuts.",
    "Inbox": "A digital abyss where emails go to be dealt with later, which everyone knows means never.",
    "Adulthood": "The prolonged realization that nobody actually knows what they are doing accompanied by an inexplicable interest in weather forecasts and comfortable shoes."
}
    
```

<i>2) Write a simple function that, when run with no arguments, outputs a random key-value pair from your heenen dictionary; format the output into a sentence that makes some sort of sense (e.g., "The definition of X is...").  Don't worry about proper capitalization or anything like that, though.

```python
#2
import random
def heenen_definition():
    word = random.choice(list(heenen.keys()))
    definition = heenen[word]
    print(f"The definition of {word} is: {definition}")

heenen_definition()
```
