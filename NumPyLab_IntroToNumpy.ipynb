<b>Part 2: Arrays and NumPy

Let's continue by jumping right into NumPy (Numerical Python) arrays. A few things to note:
1. NumPy has all of the same functions as the math module (np.sin(), np.cos(), np.pi, np.e, etc), and more.  This gives us lots of ability to work with arrays very quickly!
2. To use the NumPy module, we will must import it (like many other modules and libraries).  In addition, we will alias it (see below).

```python
# To use NumPy, use the following line at the top of your code:
import numpy as np # np is now the alias for NumPy, so we could write np.cos() instead of numpy.cos()
```

NumPy has a ton of functionality for many aspects of programming. One of its most useful features is the NumPy array. NumPy arrays can be used to store data rather like in a spreadsheet or grid. As noted above, NumPy arrays can have multiple dimensions. The most common are 1-D and 2-D arrays, though arrays can be larger.  

Every element within a NumPy array must have the same data type (what we've talked about as floats, integers, strings, etc.)  Datatypes include:
1. np.float64
2. np.int64
3. np.complex128
4. np.bool_  

The datatype will be determined automatically upon creation unless you tell it otherwise; if the elements are all entered as integers, NumPy will assume you want an integer array.

<u>Creating NumPy Arrays</u>

There are many ways to create NumPy arrays. The most common is the array() function. This function takes the argument as a list or tuple.

```python
# Example:
a1 = np.array([1, 2, 3 ,4, 5])
a2 = np.array((1., 2., 3., 4., 5))
print(a1)
print(a2)
```

Note how the output isn't separated by commas when you print the output.  Also, note that the output is surrounded by brackets but isn't a list.  We can check that using the type() function we've used before:

```python
type(a1)
```

Like most variables, Python will return the contents of the array if you just type it at the command line:

```python
a2
```

I prefer explicitly typing an array name without the print statement to see what's in it since it clearly reminds me that I'm looking at an array.  

Note the difference in data type between a1 and a2. The first array is composed of integers, while the second array is composed of floating point values. We can use the optional dtype argument to force the data type for our array:

```python
#Example
b1 = np.array([1, 2, 3 ,4, 5], dtype=np.int64)
b2 = np.array([6., 7., 8., 9., 10.], dtype=np.int64)
print(b1, b2)
```

Note that despite the fact that our list in b2 is floating point values, the dtype argument changed it to integers.

We can look at two-dimensional arrays as well.  A 2-D array can add a lot of functionality, but we have to make sure we can reference the right points within the array:

```python
# 2-D array:
c1 = np.array( ([1, 2, 3], [9, 8, 7]) )
print(c1)
```

Note how this array was created &ndash; each row was set up as a list inside of a larger tuple.  Both could be lists or tuples, but the idea for building an array this way is to have nested lists/tuples for each dimension.

<u>More Ways to Initialize Arrays

Sometimes we want to start with an array that is filled with all ones or all zeros; we can then populate the array with data later.  To do this, we can use the np.zeros() or np.ones() function. These will create an array based on a given shape and fill the array with zeros or ones. As before, the shape looks like: (Rows, Columns)

```python
arr1 = np.zeros( (1, 5) ) # 1 row, 5 columns
arr1
```

```python
arr2 = np.ones( (7, 3) ) # 7 rows, 3 columns
arr2
```
