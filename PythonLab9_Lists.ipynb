<i>7) Use a function to parse through a list of NBA team nicknames (Timberwolves, Celtics, Knicks, etc.).  Your goal is to determine how many nicknames have their last letter <u>NOT</u> as an "s".  You should also create a new list (via the append() method) that creates a new list of team nicknames that don't end with an s.  Have the function return a list containing those team names.  
    
<i>The full list of NBA teams is provided for you in the next cell; answer in the cell below that.  

```python
# Run this to create the NBA nickname list:
nba=["76ers","Bucks","Bulls","Cavaliers","Celtics","Clippers","Grizzlies","Hawks","Heat","Hornets","Jazz","Kings","Knicks","Lakers","Magic","Mavericks","Nets","Nuggets","Pacers","Pelicans","Pistons","Raptors","Rockets","Spurs","Suns","Thunder","Timberwolves","Trail Blazers","Warriors","Wizards"]
```

```python
#7
def not_an_s(inlist):
    no_s=[]  # Start empty list...
       # This is the part you'll need to figure out.  I recommend a for loop, like "for x in y"...
    for x in inlist:
        if x[-1] != "s":
            no_s.append(x)
            
    return no_s

nba_no_s = not_an_s(nba)
print(nba_no_s)
```
