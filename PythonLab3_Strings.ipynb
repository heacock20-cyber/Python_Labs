### Part 1: String Theories

Today, we're going to cover strings.  Remember that strings don't just mean "not numbers" &ndash; think of them as literal text, which can be numbers, letters, spaces, and the like.  They're important enough that they get their own format and rules!

Strings can be enclosed in single or double quotes.  Python doesn't care as long as the opening and closing quote are the same style.  This can be helpful if you want to use a single quote as an apostrophe:
- "I'm going home!" works fine as one single string
- 'I'm going home!' would confuse Python, since it would think 'I' is the first string, then other code/variables follow, then a new string starts with the last single quote

I usually default to double quotes for strings because of this effect, but you can use either.
