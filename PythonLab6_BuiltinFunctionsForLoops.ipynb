### Part 2: For Loops

The text book introduced while loops first, which is somewhat unconventional, but we will formally introduce for loops here and show more throughout the semester.  

<u>Make sure you take time to understand what's going on with for loops.  We'll be using structures like this a LOT.</u>

A for loop can work for all sorts of things, including ranges, lists, and strings, and more.  Let's stick with ranges for now.

Let's put the last range example into a for loop.  Check out and run the following cell:

```python
for num in range(1,9,2):
    print(num)
```

When you run this simple code, you get printed output from each step.  Note that the for loop structure is similar to that of an if statement &ndash; it ends with a colon and the code within the loop is indented.  Also note that Python is doing its typical weird thing with math, including the first number in the range but cutting off before it gets to the ending number.

However, like many Python functions, range doesn't need all three arguments to run.  This can be very tricky if you're not sure what Python is doing.  Let's parse through the following range using a for loop to see what happens at each level.

```python
for num in range(1,5,1):
    print(num)
```

This should print out each number from 1 to 4 on its own line.  Python will go through each element of the range and print the appropriate value.

However, if we want the step to be 1 (the default), we don't have to include it in the arguments to the range function call:

```python
for num in range(1,5):
    print(num)
```

We can go even simpler!  If we want to start at 0 (like typical Python counting), we only need to include the stop number:

```python
for num in range(5):
    print(num)
```

Note that this is different from how we started &ndash; Python starts counting at 0!  I recommend you check each range you make to ensure it's doing what you expect.  It's easy enough to define a range in a new cell to make sure it's what you want, then copy that code into your cell.

<i>6) Using the variable with your name that you set up in #2, start a for loop over a string.
    
<i>In this for loop, you're going to have an if statement embedded within.  In this case:
- <i>if the letter is capitalized, I want your print statement to note that fact
- <i>if the letter is lowercase (or it's a space), do nothing special

<i>So, if the letter is "b", have Python print "b".  However, if the letter is "G", have Python print "G <-- this letter is capitalized!  That's pretty neat!"   

```python
#6
fullname = "Abigail Heacock"

len(fullname)

for l in fullname:
    if l.isupper():
        print(f"{l} <-- this letter is capitalized! That's pretty neat!")
    else:
        print(l)
```
