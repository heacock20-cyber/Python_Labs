What you noted in the problem above is that Jupyter is currently running this cell.  A running cell means that an instance of a Python kernel is running.  A kernel in this case is basically a single instance of Python set up for just that notebook at that time.  If you only have this one notebook open, you have one Python kernel running, waiting to receive instructions.  If you have three notebooks open at the same time, you will have three different kernels running.  Think of this like having three different programs running on your computer &ndash; each kernel is its own, independent process.  

You can check to see how many kernels you currently have running by clicking on the circle with a square in it to the left of the file tree.  I don't use this often (I scroll through to see what's green instead), but it's a nice feature to know about if you have many things running.

Many of the options in the menu bar at the top are self-explanatory, so we won't go through all of them.  

The Edit menu contains some particularly useful selections:
- Split Cell will break your current cell into two different cells at the location of the cursor
- Merge Cell combines the selected cell with the cell above or below
- Move Cell is rather obvious, but can be helpful as you start writing your own code

Let's next focus on the Kernel dropdown menu at the top.  Here's what the menu contains, along with basic definitions:
- Interrupt: stop the running command (same as the stop button in the button menu)
- Restart: restart the kernel.  Useful to try if something gets stuck or of your kernel times out (shuts down because idle)
- Restart & Clear Output: restart the kernel and clear the results of all cells
- Restart Kernel and Run All Cells: restart the kernel and run everything from top to bottom
- Restart Kernel and Run up to Selected Cell: restart the kernel and run until the selected cell
- Reconnect: reset connection to a kernel
- Shutdown: stop the kernel from running
- Change kernel: on some versions of Jupyter, multiple kernels for other languages are available.  Our system only has Python 3, but more than 40 languages are now available

We'll find many of these useful, particularly Interrupt today.  So far, we haven't had a command that will cause the system to "hang" &ndash; either run out of memory or get stuck in an infinite loop.  Later, we'll test this by running bad code!

Your last problem for today is a bit more complicated.  Some of you are likely familiar with the Fibonacci Sequence; it shows up a lot in math and in nature.  Basically, the next number in the sequence is the sum of the previous two, like so:

<i>7) Your task is to write a Fibonacci sequence code that starts with the 0 and 1 and calculates the Fibonacci numbers closest to a desired number without going over.  For example, you should be able to answer the question "What's the greatest Fibonacci number under 10?" with the answer 8.
    
<i>Your code should:
- <i>use a while loop and variable reassignment to do all of this for you.  
- <i>take only one input (in our example, the 10).  
- <i>print the value of the correct number.  
    
<i>Write your function in the answer cell below, then prove that your code works by running the two cells below. 

```python
#7
n = int(input("What's the greatest Fibonacci number under 10?"))

while n <= 10 and n >= 0:
    if n == 8:
        print("8 is the answer")
        break
    else:
        print("Try again")
    n = int(input("What's the greatest Fibonacci number under 10?"))
```
