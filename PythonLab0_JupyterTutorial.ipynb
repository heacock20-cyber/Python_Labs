Jupyter Notebook is an interactive coding environment that can be used for many programming languages. Jupyter originally supported Julia, Python, and R (which is where the name comes from), but now supports over 40 programming languages.

You may occasionally find references to IPython Notebooks &ndash; that was a precursor to Jupyter.  Filenames saved in Jupyter Notebook format still hold that history, with a .ipynb extension.

Each Jupyter Notebook runs using a kernel, which is basically a single instance of the programming language run by the notebook.  On our systems, we have only have the kernel for Python 3 installed.  When you open a new notebook, you'll choose a kernel to use.  This notebook is already set to use the Python 3 kernel. There are lots of things about Jupyter Notebooks that are useful (and confusing!)  For now, you need to know that there are two main "modes" for working within a Jupyter Notebook &ndash; Command Mode and Edit Mode.  Edit Mode is fairly self-explanatory, and the keyboard shortcuts for Edit Mode are fairly similar to a standard word processor (like CTRL-C to copy).  

There are two ways to get into Edit Mode:
- with a mouse, click or double-click on a cell (depending on the cell type)
- with a keyboard, press Enter when the cell is highlighted

Command Mode is extremely useful for moving around the notebook quickly using your keyboard.  To get to Command Mode, press **Esc**.  Learning some of these keyboard shortcuts will make you much more efficient with your code.  Here are some of the most useful:

* **a** will create a cell above the current cell
* **b** will create a cell below the current cell
* **d,d** (pressing d twice) will delete the current cell
* **m** will change the cell type to markdown
* **y** will change the cell type to code
* **x, c, v** will cut, copy, and paste cells, respectively
* **Ctrl-s** will save your work.  Notebooks should autosave after some time, but it's always good to save!  The system is also set to log you out after a certain amount of idle time, and you don't want to lose your work.
* **Ctrl-Shift-Minus** will split the current cell at the location of the cursor
* **Ctrl-Shift-h** will bring up the full help list for these keyboard shortcuts

We need to talk briefly about the kernel itself.  The kernel is the computer program that's processing your Python code.  In the upper right of the notebook window, you'll see a circle called the kernel indicator.  When it's empty, the kernel is idle; when it's filled, the kernel is processing.  You won't see it filled in today (the code is too simple), but as we get further into the semester, you'll see that circle fill up, indicating that Python is processing something for you.  

One last note: whenever you close your notebook, any variables you created last time are no longer stored, even if they are still appearing in the output.  You will need to re-execute your code blocks to recreate all your variables when you re-open a notebook.  You can also either click the fast-forward button at the top of the window or go to the **Kernel** menu and select **Restart Kernel and Run All Cells**.
