### Part 3: All the Print That's Fit to Print

You've already used and seen the Python command print() a couple times.  If you give print one argument (the thing in parentheses), Python will print it:

```python
print("Hello!")
```

What if you want to print more than one thing?  There are two ways to do this in Python; you might find one or the other more useful depending on your situation.

If you separate the elements you want to print using commas, Python will output each element one after the other with a single space between.  You don't get to control the spacing this way, but you can mix different variable types, like the following:

```python
print("The answer to the meaning of life is", 42)
```

While this is great, one can run into issues if you want to form proper sentences.  If we tried to add a period at the end (using another string), we'd get a space between the number and the string:

```python
print("The answer to the meaning of life is", 42, ".")
```

To fix this issue, we can use Python's other method of printing &ndash; string concatenation.  If you put a plus sign between two strings, Python will assume you want to stick them end to end with no spaces in between.  This gives you a whole lot more control:

```python
print("If"+"you"+"don't"+"include"+"spaces"+"it"+"sounds"+"like"+"you've"+"had"+"too"+"much"+"coffee!")
```

However, for this method to work, you'll have to convert any number to a string (Python can't concatenate floats, integers, or booleans).  To do this, use the str() function:

```python
print("My favorite number is "+str(-1))
```
