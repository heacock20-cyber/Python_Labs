### Part 2: Writing Your Own Functions

The textbook does a pretty good job of introducing basic functions, but it's not the clearest about how functions actually work.  Let's look at a couple examples to help clarify things.

Here's the basic layout of a function.  Run the following cell to see the output, then we'll figure out what each part is.

```python
def functionname(argument):
    print("Anything indented will be considered part of the function.")
    print("Your function was called using: "+argument)

functionname("foo")
```

So, what happened?
- The def line is where Python defines a function.  Once defined, a function can be used as many times later as you'd like.  
- The argument is what's in the parentheses after the name of the function.  It's a bit tricky, but think of it as the name of a variable you can use only within the function.  When you run the function (the bottom line in the cell), the argument's value is changed to what you entered (here, "foo").
- Everything within the function is indented.  Python will know the function is done when the indentation stops.

Note that we can run the function (once defined) in any other cell, like:

```python
functionname('bar')
```

We can also make functions that do more than print things.  Here's a simple math-based example:

```python
def pythagoras(a,b):
    c=math.sqrt(a**2+b**2)
    print("Your hypotenuse length is:")
    print(c)
    
pythagoras(3,4)
```

You'll notice that this function takes two arguments (a and b) and needs them to be numbers, not strings.  We can write code to check to make sure the inputs are numbers, but that's a much later problem.

Note also what happens in the code cell below; you should get an error:

```python
print(c)
```

You get this error because c is considered a local variable &ndash; it's only defined within the pythagoras function but not in the main Python program overall.  This can trip you up if you're not careful!  There are ways to output variables from functions using <i>return</i>, but we'll cover that later.
